```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,5))
plt.hist(data["stars"], bins=30, color='blue', alpha=0.7)
plt.title('Distribution of GitHub Repository Stars')
plt.xlabel("stars")
plt.ylabel("number of repositories")
plt.show() 
```

```python
plt.figure(figsize=(8,5))
plt.hist(data["forks"], bins=20)
plt.title('Distribution of GitHub Repository Forks')
plt.xlabel("forks")
plt.ylabel("number of repositories")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(data["watchers"], bins=20)
plt.title('Distribution of GitHub Repository Watchers')
plt.xlabel("watchers")
plt.ylabel("number of repositories")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(data["stars"], data["forks"])
plt.title('Relationship between Stars and Forks')
plt.xlabel("stars")
plt.ylabel("forks")
plt.show()
```

```python
yearly_repos=data.groupby(data["created_at"].dt.year).size()
```

```python
plt.figure(figsize=(10,5))
plt.plot(yearly_repos.index, yearly_repos.values, marker='o')
plt.title('Yearly Distribution of GitHub Repositories')
plt.xlabel('Year')
plt.ylabel('Number of Repositories')
plt.grid(True)
plt.show()
```

```python
top10_stars=data.nlargest(10,"stars")
plt.figure(figsize=(10,6))
plt.bar(top10_stars["repo_name"],top10_stars["stars"])
plt.title('Top 10 GitHub Repositories by Stars')
plt.xlabel('Repository Name')
plt.ylabel('Number of Stars')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
language_count=data["language"].value_counts().head(10)
```

```python
plt.figure(figsize=(10,6))
plt.bar(language_count.index, language_count.values)
plt.title('Top 10 Programming Languages in GitHub Repositories')
plt.xlabel('Programming Language')
plt.ylabel('Number of Repositories')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(data["community_health_percentage"], bins=20)
plt.title('Distribution of Community Health Percentage')
plt.xlabel('Community Health Percentage')
plt.ylabel('number of repositories')
plt.show()
```
